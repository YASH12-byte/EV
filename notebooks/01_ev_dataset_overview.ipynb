```python
regs_path = PROC / "ev_registrations_annual.csv"
if regs_path.exists():
    regs = pd.read_csv(regs_path)
    display(regs.head())
    print(regs.groupby("Year")["Registrations"].sum().tail(10))
else:
    print("Missing ev_registrations_annual.csv — run: python run_project.py")
```

```python
import sys
sys.path.insert(0, str(ROOT))
from src.state_names import display_name

if regs_path.exists():
    top = (
        regs.groupby("State", as_index=False)["Registrations"]
        .sum()
        .sort_values("Registrations", ascending=False)
        .head(10)
    )
    top["StateName"] = top["State"].map(display_name)
    display(top[["StateName", "Registrations"]])
```
